```python

import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer


class TextClassificationDataset(Dataset):
    def __init__(self, questions, answers, labels, tokenizer, max_length=256):
        self.questions = list(questions)
        self.answers = list(answers)
        self.labels = list(labels)
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            self.questions[idx],
            self.answers[idx],
            truncation=True,
            padding="max_length",
            max_length=self.max_length,
            return_tensors="pt"
        )

        return {
            "input_ids": encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels": torch.tensor(self.labels[idx], dtype=torch.long)
        }


# Exemplu independent de datele competitiei
questions = [
    "Cand va fi gata proiectul?",
    "Cine este responsabil?"
]

answers = [
    "Proiectul va fi gata in martie.",
    "Nu cunosc aceasta informatie."
]

labels = [0, 1]

tokenizer = AutoTokenizer.from_pretrained(
    "FacebookAI/xlm-roberta-base"
)

dataset = TextClassificationDataset(
    questions,
    answers,
    labels,
    tokenizer
)

loader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=True
)

for batch in loader:
    print("input_ids:", batch["input_ids"].shape)
    print("attention_mask:", batch["attention_mask"].shape)
    print("labels:", batch["labels"])
```

```python
import pandas as pd

df = pd.read_csv("/content/train_data.csv")

print("Dataset shape:", df.shape)
print("\nDistribuția claselor:")
print(df["evasion_label"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

labels = sorted(df["evasion_label"].unique())

label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}

df["label_id"] = df["evasion_label"].map(label2id)

train_df, val_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["label_id"]
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Classes:", label2id)
```

```python
from torch.utils.data import DataLoader

train_dataset = TextClassificationDataset(
    train_df["interview_question"],
    train_df["interview_answer"],
    train_df["label_id"],
    tokenizer,
    max_length=256
)

val_dataset = TextClassificationDataset(
    val_df["interview_question"],
    val_df["interview_answer"],
    val_df["label_id"],
    tokenizer,
    max_length=256
)

train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=4, shuffle=False)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
```

```python
from transformers import AutoModelForSequenceClassification
from torch.optim import AdamW

model = AutoModelForSequenceClassification.from_pretrained(
    "FacebookAI/xlm-roberta-base",
    num_labels=9,
    id2label=id2label,
    label2id=label2id
)

model = model.to(device)

optimizer = AdamW(model.parameters(), lr=2e-5)
```

```python
from sklearn.metrics import accuracy_score, f1_score
import torch

def evaluate_model(model, loader, device):
    model.eval()

    y_true = []
    y_pred = []

    with torch.no_grad():
        for batch in loader:
            batch = {
                key: value.to(device)
                for key, value in batch.items()
            }

            outputs = model(
                input_ids=batch["input_ids"],
                attention_mask=batch["attention_mask"]
            )

            predictions = outputs.logits.argmax(dim=1)

            y_true.extend(batch["labels"].cpu().tolist())
            y_pred.extend(predictions.cpu().tolist())

    accuracy = accuracy_score(y_true, y_pred)
    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=list(range(9)),
        average="macro",
        zero_division=0
    )

    return accuracy, macro_f1
```
